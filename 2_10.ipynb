```python
from time import sleep
from IPython.display import clear_output

saldo = 4200
pocket = 0

while True:
  print(f"--- Your Account (Saldo: {saldo} Kr) ---\n")
  print("[1] Put in Money")
  print("[2] Take Money Out\n")
  print(f"What do you wish to do? (You have {pocket} money in your hand)")
  option = input("> ")

  if option == "1":
    if pocket == 0:
      print("Error: No Money to put in...\n")
      sleep(1)
      clear_output()
    else:
      print(f"You have {pocket} money in your hand. How much would you like to put in?")
      put = int(input("> "))
      if put > saldo:
        print("Error: u broke :(\n")
        sleep(1)
        clear_output()
      else:
        saldo = put + saldo
        sleep(1)
        clear_output()
  elif option == "2":
    print("How much money would you like to pocket? (0.1% Exchange rate)")
    take = int(input("> "))
    if take > saldo:
      print("Error: u broke :(\n")
      sleep(1)
      clear_output()
    else:
      take = take * 0.90
      pocket = pocket + take
      saldo = saldo - pocket
      sleep(1)
      clear_output()
```
