```python
import random


Äpple = 7
Päron = 13


Vinst_P = 7*Päron
Vinst_A = 13*Äpple


print("I omgång 1 blev resulatet att:")
if Vinst_P > Vinst_A:
  print(f"Petra sålde mest med {Vinst_P} Kr!! :)")
elif Vinst_P < Vinst_A:
  print(f"Axel sålde mest med {Vinst_A} Kr!! :)")
elif Vinst_P == Vinst_A:
  print(f"De sålde lika mycket. Det blev oavgjort med {Vinst_P} Kr vardera :(\n")

Vinst_P = 8*Päron
Vinst_A = 15*Äpple

print("I omgång 2 blev resulatet att:")
if Vinst_P > Vinst_A:
  print(f"Petra sålde mest med {Vinst_P} Kr!! :)")
elif Vinst_P < Vinst_A:
  print(f"Axel sålde mest med {Vinst_A} Kr!! :)")
elif Vinst_P == Vinst_A:
  print(f"De sålde lika mycket. Det blev oavgjort med {Vinst_A} Kr vardera :(")

Vinst_P = random.randint(1,10)*Päron
Vinst_A = random.randint(1,20)*Äpple


print("\nI omgång 3 blev resulatet att:")
if Vinst_P > Vinst_A:
  print(f"Petra sålde mest med {Vinst_P} Kr!! :)")
elif Vinst_P < Vinst_A:
  print(f"Axel sålde mest med {Vinst_A} Kr!! :)")
elif Vinst_P == Vinst_A:
  print(f"De sålde lika mycket. Det blev oavgjort med {Vinst_A} Kr vardera :(")

Vinst_P = int(Päron/1)
Vinst_A = int(Äpple/1)

print("\nI omgång 4 blev resulatet att:")
if Vinst_P > Vinst_A:
  print(f"Petra sålde mest med {Vinst_P} Kr!! :)")
elif Vinst_P < Vinst_A:
  print(f"Axel sålde mest med {Vinst_A} Kr!! :)")
elif Vinst_P == Vinst_A:
  print(f"De sålde lika mycket. Det blev oavgjort med {Vinst_A} Kr vardera :(")
```
